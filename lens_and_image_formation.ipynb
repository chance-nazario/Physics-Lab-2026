```python
import numpy as np
import matplotlib.pyplot as plt

# ==========================================
# PART 1: Converging Lens
# ==========================================
x_1 = np.array([77.4, 59.7, 64, 71.5, 81.4])
y_1 = np.array([1148, 891, 960, 1075, 1284])
dy_1 = np.array([17.2, 11.9, 13.2, 15.4, 18.1])

# Part 1 Fit and Error
b_1, m_1 = np.polynomial.polynomial.polyfit(x_1, y_1, 1, w=dy_1)
fit_1 = b_1 + m_1 * x_1

D_1 = (sum(1/dy_1**2))*(sum(x_1**2/dy_1**2)) - (sum(x_1/dy_1**2))**2
dm_1 = np.sqrt(1/D_1 * sum(1/dy_1**2))
db_1 = np.sqrt(1/D_1 * sum(x_1**2/dy_1**2))

# ==========================================
# PART 2: Converging and Diverging Lens
# ==========================================
x_3 = np.array([33.6, 16.6, 8, 4.6, 0.9])
y_3 = np.array([-703.72, -367.92, -197.16, -113.52, -49.5])
dy_3 = np.array([14.3, 9.0, 6.1, 4.5, 2.9])

# Part 2 Fit and Error
b_3, m_3 = np.polynomial.polynomial.polyfit(x_3, y_3, 1, w=dy_3)
fit_3 = b_3 + m_3 * x_3

D_3 = (sum(1/dy_3**2))*(sum(x_3**2/dy_3**2)) - (sum(x_3/dy_3**2))**2
dm_3 = np.sqrt(1/D_3 * sum(1/dy_3**2))
db_3 = np.sqrt(1/D_3 * sum(x_3**2/dy_3**2))

# ==========================================
# PLOTTING LOGIC (Side-by-Side Figure)
# ==========================================
# Create a figure with 1 row and 2 columns
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(18, 8))

# --- Plot 1: Force vs Displacement ---
ax1.plot(x_1, fit_1, color='green', linestyle='--')
ax1.scatter(x_1, y_1, color='blue', marker='o')
ax1.errorbar(x_1, y_1, yerr=dy_1, xerr=None, fmt="none", color='black')

ax1.set_xlabel('Do + Di (cm)')
ax1.set_ylabel('Do*Di (cm^2)')
ax1.set_title('Do*Di vs Do + Di')

# Combine annotations into a clean text box for ax1
stats_1 = (f'Slope (cm^2/cm) = {m_1:.7f}\n'
           f'Error in Slope = {dm_1:.7f}\n'
           f'Vertical intercept = {b_1:.7f}\n'
           f'Error in intercept = {db_1:.7f}')
ax1.annotate(stats_1, (0.45, 0.80), xycoords='axes fraction',
             bbox=dict(boxstyle="round,pad=0.3", fc="white", ec="gray", lw=1))

# --- Plot 3: Displacement vs Current ---
ax2.plot(x_3, fit_3, color='green', linestyle='--')
ax2.scatter(x_3, y_3, color='blue', marker='o')
ax2.errorbar(x_3, y_3, yerr=dy_3, xerr=None, fmt="none", color='black')

ax2.set_xlabel('Do + Di (cm)')
ax2.set_ylabel('Do*Di (cm^2)')
ax2.set_title('Do*Di vs Do + Di')

# Combine annotations into a clean text box for ax2
stats_3 = (f'Slope (cm^2/cm) = {m_3:.7f}\n'
           f'Error in Slope = {dm_3:.7f}\n'
           f'Vertical intercept = {b_3:.7f}\n'
           f'Error in intercept = {db_3:.7f}')
ax2.annotate(stats_3, (0.45, 0.80), xycoords='axes fraction',
             bbox=dict(boxstyle="round,pad=0.3", fc="white", ec="gray", lw=1))

# Adjust layout to prevent overlap and display the figure
plt.tight_layout()
plt.show()
```
